# CulSen-BN: 07. Confidence-Filtered Supervised Contrastive Learning (SupCon)

Projects pooled representations into a normalized sentiment metric space (128-d) and filters ambiguous pseudo-labels.


```python
import sys
from pathlib import Path
sys.path.append(str(Path(".").resolve()))
sys.path.append(str(Path("src").resolve()))

import torch
from culsen_bn.models.contrastive import SupervisedContrastiveHead, SupConLoss

proj_head = SupervisedContrastiveHead(input_dim=768, proj_dim=128)
supcon_loss = SupConLoss(temperature=0.07, confidence_threshold=0.65)

features = torch.randn(8, 768)
labels = torch.tensor([1, 1, 0, 0, 1, 0, 1, 0])
conf = torch.tensor([0.9, 0.8, 0.7, 0.6, 0.95, 0.85, 0.5, 0.9])

z = proj_head(features)
loss = supcon_loss(z, labels, confidence=conf)
print("SupCon Loss value:", loss.item())
```
